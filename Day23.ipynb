```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the CSV with correct delimiter and quoting
car = pd.read_csv("mpg.csv", quotechar='"', skipinitialspace=True)

# Normalize column names
car.columns = car.columns.str.strip().str.lower()

# Confirm column names
print(car.columns.tolist())  # 'cylinders' should be in this list
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Define correct column names manually
columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight',
           'acceleration', 'model_year', 'origin', 'name']

# Load the CSV and assign column names
car = pd.read_csv("mpg.csv", names=columns, header=None)

# Convert 'cylinders' to numeric
car['cylinders'] = pd.to_numeric(car['cylinders'], errors='coerce')

# Drop missing values
car_clean = car.dropna(subset=['cylinders'])

# Plot KDE
sns.kdeplot(data=car_clean, x='cylinders', fill=True)
plt.title("KDE of Car Cylinders")
plt.xlabel("Cylinders")
plt.ylabel("Density")
plt.show()
```
